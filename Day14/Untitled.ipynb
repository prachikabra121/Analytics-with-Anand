
<h2>1. Shallow Copy</h2>
A shallow copy creates a new object but does not create copies of nested objects (e.g., lists within a list). Instead, it copies references to the nested objects.

```python
import copy
```

```python
original = [[1, 2, 3], [4, 5, 6]]
```

```python
shallow_copy=copy.copy(original)
```

```python
shallow_copy[0][0]=99
```

```python
shallow_copy
```

```python
original
```

```python
shallow_copy[0]=[0,0,0]
```

```python
shallow_copy
```

```python
original
```

```python
id(original)
```

```python
id(shallow_copy)
```

```python
import copy

# Original object with nested list
original = [[1, 2, 3], [4, 5, 6]]
```

```python
Key Points:
The outer structure (list) is copied.

Nested objects (inner lists) are referenced, not copied.

Changes to nested objects in the shallow copy affect the original object.
```

2. Deep Copy
A deep copy creates a new obj`ect and recursively copies all nested objects, ensuring complete independence from the original object.

```python
original = [[1, 2, 3], [4, 5, 6]]
```

```python
deep_cpy=copy.deepcopy(original)
```

```python
original
```

```python
deep_cpy
```

```python
id(deep_cpy)
```

```python
id(original)
```

```python
id(deep_cpy[0])
```

```python
id(original[0])
```

```python
deep_cpy[0][0]=99
```

```python
deep_cpy
```

```python
original[0][1]=98
```
